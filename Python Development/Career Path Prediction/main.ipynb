# Career Path Predictor

This beginner project uses a small sample dataset of skills, interests, and background information to suggest one of six career categories. Run the cells from top to bottom to train the model and try a sample profile.

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.tree import DecisionTreeClassifier
```

```python
data = pd.read_csv("dataset/data.csv")
print(f"Sample profiles: {len(data)}")
print("Examples per career category:")
print(data["Target_Career"].value_counts())
data.head()
```

```python
numeric_features = [
    "Logical_Thinking",
    "Coding_Skills",
    "Public_Speaking",
    "Creativity",
]
categorical_features = ["Preferred_Environment", "Academic_Background"]
features = numeric_features + categorical_features

preprocessor = ColumnTransformer(
    transformers=[
        ("categories", OneHotEncoder(handle_unknown="ignore"), categorical_features),
        ("scores", "passthrough", numeric_features),
    ]
)
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("decision_tree", DecisionTreeClassifier(max_depth=5, random_state=42)),
    ]
)
model.fit(data[features], data["Target_Career"])
print("Model trained on", len(data), "sample profiles.")
```

```python
new_person = pd.DataFrame(
    [
        {
            "Logical_Thinking": 9,
            "Coding_Skills": 9,
            "Public_Speaking": 3,
            "Creativity": 4,
            "Preferred_Environment": "Remote",
            "Academic_Background": "Computer Science",
        }
    ]
)
prediction = model.predict(new_person)[0]
print("Suggested career category:", prediction)
```

## How the model works

Each sample row pairs four 1-to-10 skill ratings and two background choices with a career label. The notebook converts the text choices into numeric features, then trains a Decision Tree to learn patterns in those examples. For a new profile, the tree follows learned decision rules and returns the closest career category. This tiny teaching dataset is illustrative, so the prediction is a suggestion, not career advice.
